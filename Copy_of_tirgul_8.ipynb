```python
import time
import random

class PlantManagementService:
    """ניהול ישויות הצמחים במערכת"""
    def __init__(self):
        self.plants = {
            'plant_1': {'id': 'plant_1', 'name': 'Tomato Alpha', 'type': 'Tomato', 'user_id': 'user_101'},
            'plant_2': {'id': 'plant_2', 'name': 'Basil Bravo', 'type': 'Basil', 'user_id': 'user_102'}
        }

    def get_plant(self, plant_id):
        return self.plants.get(plant_id, {})


class IoTDataIngestionService:
    """שירות קליטת נתוני חיישנים בזמן אמת מהשטח"""
    def __init__(self):
        self.telemetry_db = []

    def ingest_metrics(self, event):
        start_time = time.time()

        # סימולציית שמירה קלה ב-Time-Series DB
        self.telemetry_db.append(event)

        latency = (time.time() - start_time) * 1000 # המרה למיל שניות
        return {
            'status': 'success',
            'latency_ms': latency,
            'sla_passed': latency < 500 # יעד ה-SLA הוא פחות מ-500ms
        }


class AIAnalysisService:
    """שירות מבוסס AI המנתח תמונות צמחים לזיהוי מחלות ומזיקים"""
    def __init__(self):
        self.analysis_logs = {}

    def analyze_image(self, event):
        start_time = time.time()

        # סימולציה של עיבוד תמונה כבד (לוקח בין 0.5 ל-1.5 שניות)
        time.sleep(random.uniform(0.5, 1.5))

        doc_id = event['document_id']
        content = event['content']

        # הדמיית זיהוי מבוסס מילים קריטיות בטקסט/תמונה
        detected_issues = []
        if 'yellow' in content.lower():
            detected_issues.append('Yellow Leaves Symptom')
        if 'spots' in content.lower():
            detected_issues.append('Fungal Spots')
        if 'dry' in content.lower():
            detected_issues.append('Under-watering')

        latency = time.time() - start_time

        return {
            'status': 'success',
            'doc_id': doc_id,
            'detected_issues': detected_issues,
            'latency_seconds': latency,
            'sla_passed': latency < 3.0 # יעד ה-SLA הוא פחות מ-3 שניות
        }
```

**סעיף 3**

```python
class PlantIndexService:
    """מנוע אינדקס הפוך עם תמיכה בדירוג (Ranking) עבור מדריכי טיפול ומחלות צמחים"""
    def __init__(self):
        self.documents = {}
        self.index = {}  # מילה -> קבוצה של מזהי מסמכים

    def add_document(self, doc_id, title, content):
        """הוספת מסמך/מדריך אגרונומי לאינדקס"""
        self.documents[doc_id] = {'id': doc_id, 'title': title, 'content': content}

        words = content.lower().replace('.', '').replace(',', '').split()
        for word in words:
            if word not in self.index:
                self.index[word] = set()
            self.index[word].add(doc_id)

    def _score_document(self, doc_id, terms):
        """מנגנון דירוג פשוט (Ranking) - סופר כמה פעמים מילות החיפוש מופיעות במסמך"""
        doc = self.documents.get(doc_id)
        if not doc:
            return 0
        content_lower = doc['content'].lower()
        score = 0
        for term in terms:
            score += content_lower.count(term.lower())
        return score

    def search_advanced(self, query_string, operator='AND'):
        """
        מנגנון חיפוש לוגי מתקדם (AND / OR) כולל דירוג התוצאות
        """
        terms = query_string.lower().split()
        if not terms:
            return []

        # שלב 1: איסוף רשימות התוצאות לכל מילה בנפרד
        term_results = []
        for term in terms:
            term_results.append(self.index.get(term, set()))

        # שלב 2: ביצוע החיתוך הלוגי (AND או OR)
        if operator.upper() == 'AND':
            # חיתוך (Intersection) - המסמך חייב להכיל את כל המילים
            final_doc_ids = set.intersection(*term_results) if term_results else set()
        elif operator.upper() == 'OR':
            # איחוד (Union) - המסמך יכול להכיל לפחות אחת מהמילים
            final_doc_ids = set.union(*term_results) if term_results else set()
        else:
            raise ValueError("Operator must be 'AND' or 'OR'")

        # שלב 3: דירוג התוצאות (Ranking) לפי רלוונטיות
        ranked_results = []
        for doc_id in final_doc_ids:
            score = self._score_document(doc_id, terms)
            doc = self.documents[doc_id]
            ranked_results.append({
                'doc_id': doc_id,
                'title': doc['title'],
                'score': score,
                'snippet': doc['content'][:70] + "..."
            })

        # מיון מהציון הגבוה לנמוך (הכי רלוונטי בהתחלה)
        ranked_results.sort(key=lambda x: x['score'], reverse=True)
        return ranked_results
```

**סעיף 4**

```python
class GamificationEngineFunction:
    """פונקציית Serverless המטפלת באירועי גיימיפיקציה בצורה אסינכרונית"""
    def __init__(self):
        self.user_scores = {'user_101': 50, 'user_102': 120}

    def handle(self, event):
        """קבלת אירוע (למשל: המשתמש פתר בעיית השקיה) ועדכון הניקוד"""
        event_type = event.get('event_type')
        user_id = event.get('user_id')

        if event_type == 'daily_task_completed':
            self.user_scores[user_id] = self.user_scores.get(user_id, 0) + 10
            return {
                'status': 'success',
                'msg': f"Points added! New score for {user_id}: {self.user_scores[user_id]}"
            }
        return {'status': 'ignored'}


class PlantCloudFaaSSimulator:
    """ניהול וניתוב הפונקציות בענן (ה-FaaS Simulator המורחב)"""
    def __init__(self):
        self.ai_function = AIAnalysisService()
        self.iot_function = IoTDataIngestionService()
        self.gamification_function = GamificationEngineFunction()
        self.total_invocations = 0

    def invoke(self, function_name, event):
        self.total_invocations += 1
        if function_name == 'ai_analysis':
            return self.ai_function.analyze_image(event)
        elif function_name == 'iot_ingestion':
            return self.iot_function.ingest_metrics(event)
        elif function_name == 'gamification':
            return self.gamification_function.handle(event)
        else:
            raise ValueError(f"Unknown Lambda/FaaS function: {function_name}")
```

הרצה ובדיקה של כל הסעיפים יחד (Main Flow)

```python
def main():
    print("=== 1. אתחול מערכת ומנוע חיפוש (סעיף 3) ===")
    index_service = PlantIndexService()

    # הוספת מסמכי מדריך לחקלאי/מגדל
    index_service.add_document('doc1', 'Tomato Care Guide', 'Yellow leaves on tomato usually mean under-watering or lack of nitrogen.')
    index_service.add_document('doc2', 'Fungal Diseases', 'Brown spots and yellow spots on leaves indicate fungal infection.')
    index_service.add_document('doc3', 'Irrigation Basics', 'Keep soil moist. Dry soil causes stress. Over-watering causes root rot.')

    # בדיקת חיפוש לוגי מסוג OR + בדיקת מנגנון הדירוג (Ranking)
    # המילה yellow מופיעה ב-doc1 ו-doc2, אבל ב-doc2 מופיעה גם המילה spots, לכן doc2 צריך לקבל דירוג גבוה יותר עבור השאילתה הזו
    print("\n--- בדיקת חיפוש OR עם דירוג (Searching: 'yellow spots') ---")
    or_results = index_service.search_advanced("yellow spots", operator="OR")
    for res in or_results:
        print(f"Doc ID: {res['doc_id']} | Title: {res['title']} | Score: {res['score']} | Snippet: {res['snippet']}")

    # בדיקת חיפוש לוגי מסוג AND
    print("\n--- בדיקת חיפוש AND (Searching: 'yellow tomato') ---")
    and_results = index_service.search_advanced("yellow tomato", operator="AND")
    for res in and_results:
        print(f"Doc ID: {res['doc_id']} | Title: {res['title']} | Score: {res['score']}")


    print("\n=== 2. סימולציית סביבת ענן ופונקציות Serverless (סעיף 1 + 4) ===")
    cloud_env = PlantCloudFaaSSimulator()

    # א) אירוע IoT מחיישן בשטח
    print("\n--- [Event] חיישן שולח נתונים מהשטח ---")
    iot_event = {'sensor_id': 'sns_01', 'plant_id': 'plant_1', 'soil_moisture': '12%', 'temperature': '24C'}
    iot_res = cloud_env.invoke('iot_ingestion', iot_event)
    print(f"Result: {iot_res} | KPI Check (Latency < 500ms): {iot_res['sla_passed']}")

    # ב) אירוע העלאת תמונה לניתוח AI
    print("\n--- [Event] משתמש מעלה תמונת צמח עם עלים צהובים וכתומים ---")
    ai_event = {'document_id': 'upload_44', 'content': 'The plant has dry yellow leaves with some weird spots'}
    ai_res = cloud_env.invoke('ai_analysis', ai_event)
    print(f"Result: {ai_res}")
    print(f"SLA Check (Process Time: {ai_res['latency_seconds']:.2f}s) | Target < 3s: {ai_res['sla_passed']}")

    # ג) אירוע משחוק (Gamification) - המשתמש השלים משימה
    print("\n--- [Event] המשתמש ביצע משימת השקיה יומית ---")
    game_event = {'event_type': 'daily_task_completed', 'user_id': 'user_101'}
    game_res = cloud_env.invoke('gamification', game_event)
    print(f"Result: {game_res}")

    print(f"\nTotal Serverless Invocations Monitored: {cloud_env.total_invocations}")

if __name__ == "__main__":
    main()
```
